```python
import pandas as pd
import numpy as np
df_age = pd.DataFrame({
	'department' :['IT','IT','HR','HR','Sales','Sales'],
	'Salary' :[85000,92000,50000,54000,85000,70000],
 	'tenure' :[3,5,2,6,4,7]
})

print(df_age)
print("-"*60)
summary_df = df_age.groupby('department').agg(
	mean_salary = ('Salary','mean'),
	max_tenure = ('tenure','max')
).reset_index()

print("Summary :",summary_df)
```

```python
df_pivot = pd.DataFrame(
    {
        "device": [
            "Mobile","Desktop","Mobile","Desktop","Mobile","Desktop",
        ],
        "region": ["North", "North", "South", "South", "North", "South"],
        "sales": [200, 450, 300, 500, 250, 400],
    }
)

print(df_pivot)
print("-"*60)

result = pd.pivot_table(
    df_pivot,
    values="sales",
    index="device",
    columns="region",
    aggfunc="sum",
    margins=True,  # Adds 'All' row and column
)

print(result)
```

```python
df_wide = pd.DataFrame(
    {
        "product": ["Widget A", "Widget B"],
        "Jan": [100, 150],
        "Feb": [120, 170],
    }
)

df_long = df_wide.melt(id_vars=["product"])

print(df_long)
```
