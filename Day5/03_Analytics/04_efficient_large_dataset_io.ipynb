```python
import io
import pandas as pd
import numpy as np
```

```python
csv_data = "id,val\n1,10\n2,20\n3,30\n4,40\n5,50"
chunk_size = 2
total_sum = 0

df = pd.read_csv(io.StringIO(csv_data), chunksize=chunk_size)
for i, chunk in enumerate(df, start=1):
    chunk_sum = chunk["val"].sum()
    total_sum += chunk_sum
    print(f"Chunk {i} (IDs: {chunk['id'].tolist()}): sum = {chunk_sum})")
    
print(f"Total Sum: {total_sum}")
```
