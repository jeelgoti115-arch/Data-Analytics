## Automated Quality and Integrity Audits

### Definition
Automated quality audits systematically scan datasets for structural defects, including duplicate rows, invariant columns, mixed data types, out-of-range values, and broken referential constraints.

### Naive Example
Running an automated audit script that flags any table column where more than 5% of entries are missing or where categorical values violate expected domain lists.

### Uses
- Enforcing data governance and pipeline reliability checks.
- Catching ingestion anomalies before downstream model training.
- Automated data quality logging and alerting.

```python
df_audit = pd.DataFrame({
    'id': [1, 2, 2, 4, 5],
    'status': ['Active', 'Pending', 'Pending', 'Closed', None],
    'amount': [100, 200, 200, -50, 500]
})

# Audit checks
null_counts = df_audit.isnull().sum()
duplicate_count = df_audit.duplicated().sum()
print("Null Counts per Column:\n", null_counts)
print("Duplicate Rows Count:", duplicate_count)
```

## 3. Data Schema Validation with Pandas-Schema

### 1. Definition
Data schema validation enforces strict rules on DataFrame columns, data types, value ranges, and regex patterns, rejecting malformed records that fail predefined business logic contracts.

### . Naive Example
Ensuring an incoming employee dataset strictly requires `age` to be between 18 and 100 and `email` to match a valid email regex pattern.

### . Uses
- Contract testing for ETL data pipelines and API response ingestion.
- Preventing type-mismatch errors in production machine learning inference.
- Automated error reporting for failing data records.

```python
def validate_schema(df):
    errors = []
    if not pd.api.types.is_integer_dtype(df['age']):
        errors.append('Age must be integer')
    if (df['age'] < 0).any():
        errors.append('Age cannot be negative')
    return errors

df_test = pd.DataFrame({'age': [25, 30, -5, 45]})

validation_errors = validate_schema(df_test)
print("Schema Validation Errors Found:", validation_errors)
```

### high-level metadata telemetry

-  Definition
Dataset overview summaries provide high-level metadata telemetry, including total memory consumption, column data type distributions, non-null counts, and shape dimensions.

###  Naive Example
Inspecting `df.info()` immediately after loading a file to check how many columns were incorrectly imported as object strings instead of numeric floats.

###  Uses
- Quick sanity check for file loading correctness.
- Assessing memory overhead across different data types.
- Initial documentation step for analytical notebooks.

```python
df_overview = pd.DataFrame({
    'col_int': [1, 2, 3],
    'col_float': [1.1, 2.2, 3.3],
    'col_str': ['A', 'B', 'C']
})


# DataType counts summary
dtype_counts = df_overview.dtypes.astype(str).value_counts()
print("Data Type Distribution:\n", dtype_counts)
```
