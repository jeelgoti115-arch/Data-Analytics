### Bernoulli distribution

The Bernoulli distribution is a discrete probability distribution that models a single experiment (called a Bernoulli trial) which has only two possible outcomes:

• 1 (Success): occurs with probability p

• 0 (Failure): occurs with probability q = 1 - p (where 0 ≤ p ≤ 1)

### Key Formulas

* **Probability Mass Function (PMF):**

$$P(X = x) = p^x (1 - p)^{1-x} \quad \text{for } x \in \{0, 1\}.$$

* **Mean (Expected Value):**

$$\mu = p$$

* **Variance:**

$$\sigma^2 = p(1 - p) = pq$$

```python
p_bernoulli = 0.65
k_bern = np.array([0,1])
pmf_bern = stats.bernoulli.pmf(k_bern, p=p_bernoulli)

print(" 1. Bernoulli distribution")
print(f"Success parameter (p) : {p_bernoulli:.2f}")
print(f"P(X=0) Failure: {pmf_bern[0]:.4f}")
print(f"P(X=1) Success: {pmf_bern[1]:.4f}")
print(f"Theoritical mean : {stats.bernoulli.mean(p=p_bernoulli):.4f}")
print(f"Theoritical variance : {stats.bernoulli.var(p=p_bernoulli):.4f}")
```

### binomial distribution

A binomial distribution is a discrete probability distribution that models the number of successes in a fixed number of independent trials, where each trial has only two possible outcomes (commonly called "success" and "failure").

### The Binomial Formula

To find the probability of getting exactly $k$ successes in $n$ independent trials, you use the **Probability Mass Function (PMF)**:

$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}$$

```python
from scipy.stats import binom

n=5
p=0.7

prob_4 = binom.pmf(k=4,n=n,p=p)
print(f"exact probability of making 4 shots : {prob_4:.4f} (~{prob_4 * 100:.4f}%)")

simulation = binom.rvs(n=n,p=p,size=1000)
average_makes = sum(simulation) / len(simulation)
excepted_value = n * p

print(f"Average : {average_makes}")
print(f"Expected value : {excepted_value}")
```

### 4. Poisson distribution 

The Poisson distribution is a discrete probability distribution that models the exact number of times a random, independent event occurs within a fixed interval of time or space

### Key Formula

The probability of observing $k$ events in an interval is given by the probability mass function:

$$P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}$$

* **$\lambda$ (lambda):** The average (mean) number of events expected in the given interval.
* **$e$:** Euler’s constant (approximately 2.71828).
* **$k$:** The actual number of occurrences (0, 1, 2, 3, ...).
* **$k!$:** The factorial of $k$.

```python
from scipy.stats import poisson

lambda_rate = 3
prob_2 = poisson.pmf(k=2,mu=lambda_rate)
print(f"probability of getting excatly 2 orders : {prob_2:.4f} (~{prob_2*100:.1f}%)")
simulated_hours = poisson.rvs(mu=lambda_rate, size=10)
print(f"simulated orders for 10 hours (random values of x) : {simulated_hours}")
```

### Geometric Distribution

The geometric distribution models the number of independent Bernoulli trials needed to get the **first success**, where each trial has a success probability $p$.

* **Probability Mass Function (PMF)** *(Trial of first success, $k \in \{1, 2, 3, \dots\}$)*:

$$P(X = k) = (1 - p)^{k-1} p$$

* **Alternative PMF** *(Number of failures before first success, $k \in \{0, 1, 2, \dots\}$)*:

$$P(Y = k) = (1 - p)^k p$$

* **Mean (Expected Value):**

$$\mu = \frac{1}{p} \quad \text{(for trials } X\text{)} \qquad \text{or} \qquad \mu = \frac{1 - p}{p} \quad \text{(for failures } Y\text{)}$$

* **Variance:**

$$\sigma^2 = \frac{1 - p}{p^2}$$

* **Cumulative Distribution Function (CDF):**

$$P(X \le k) = 1 - (1 - p)^k$$
