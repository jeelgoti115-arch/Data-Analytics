# Discrete Probability Distributions

A **discrete random variable** $X$ is a random variable whose possible values either constitute a finite set or a countably infinite sequence (e.g., counts $0, 1, 2, \dots$). 

The behavior of $X$ is completely defined by its **Probability Mass Function (PMF)**, denoted as $P(X = k)$ or $p(k)$:

$$p(k) = P(X = k) \quad \text{where} \quad \sum_{k} p(k) = 1 \quad \text{and} \quad 0 \le p(k) \le 1$$

1. The Anatomy of a Bernoulli TrialLet's define our random variable $X$:Success (which we assign the value $1$) happens with probability $p$.Failure (which we assign the value $0$) happens with the remaining probability, which is $1 - p$.

2. The Probability Mass Function (PMF)Because there are only two outcomes, writing out the PMF is super clean. It just says:$$P(X = 1) = p \quad \text{and} \quad P(X = 0) = 1 - p$$

```python
# Import core mathematical, statistical, and plotting libraries
import matplotlib.pyplot as plt  # Scientific data visualization library
import numpy as np  # Vectorized numerical computation and array operations
from scipy import stats  # Statistical functions and continuous/discrete distributions

# Set deterministic random seed to ensure exact reproducibility across runs
np.random.seed(seed=42)

# Configure clean, readable Matplotlib typography and axis styling
plt.rcParams["font.sans-serif"] = "DejaVu Sans"
plt.rcParams["axes.edgecolor"] = "#cccccc"
plt.rcParams["axes.linewidth"] = 0.8
```

###  Bernoulli Distribution
- It represents any situation where there are only two possible outcomes.
- Coin flips, yes/no answers, passing or failing a test, clicking or not clicking an ad—these are all classic Bernoulli trials.


#### Probability Mass Function (PMF)
$$P(X = k) = p^k (1 - p)^{1 - k}, \quad k \in \{0, 1\}$$
- *Pronunciation:* "Probability that X equals k is p to the power k, times 1 minus p to the power 1 minus k, for k in 0 or 1."

#### Key Moments
- **Mean:** $\mathbb{E}[X] = p$
- **Variance:** $\text{Var}(X) = p(1 - p) = pq$ (where $q = 1 - p$)
  - *Pronunciation:* "Variance of X is p times q, where q equals 1 minus p."

```python
# Define the parameter p: probability of success for a single trial
p_bernoulli = 0.65  # 65% probability of success (e.g., automated test pass rate)

# Construct discrete outcome support vector: 0 = Failure, 1 = Success
k_bern = np.array([0, 1])  # Support set k ∈ {0, 1}

# Compute theoretical PMF values P(X = k) via scipy.stats
pmf_bern = stats.bernoulli.pmf(k_bern, p=p_bernoulli)

# Print mathematical properties and derived statistics
print("=== 1. BERNOULLI DISTRIBUTION ===")
print(f"Success Parameter (p): {p_bernoulli:.2f}")
print(f"P(X = 0) [Failure]:    {pmf_bern[0]:.4f}")  # Analytical: 1 - p = 0.3500
print(f"P(X = 1) [Success]:    {pmf_bern[1]:.4f}")  # Analytical: p = 0.6500
print(f"Theoretical Mean E[X] = p:        {stats.bernoulli.mean(p=p_bernoulli):.4f}")
print(f"Theoretical Var Var(X) = p(1-p): {stats.bernoulli.var(p=p_bernoulli):.4f}\n")
```

```python
import random

def bernoulli_trial(p):
    """
    Simulates a single Bernoulli trial with success probability p.
    Returns 1 for success, 0 for failure.
    """
    # random.random() returns a random float between 0.0 and 1.0
    if random.random() < p:
        return 1  # Success
    else:
        return 0  # Failure

# Let's test a loaded coin where the probability of success (p) is 0.6 (60%)
p_success = 0.6

# Simulate 10 trials
results = [bernoulli_trial(p_success) for _ in range(10)]
print("10 Trial Results:", results)
print("Total Successes:", sum(results))
```

## 3. Binomial Distribution
- Binomial distribution is what happens when you flip that coin multiple times and count how many total Heads you get.
- “What is the probability of getting exactly $k$ successes out of $n$ independent tries?

### The 4 Ingredients of a Binomial Situation
1. Fixed number of trials ($n$): You decide in advance how many times you're going to do the experiment (e.g., flip a coin 10 times).
2. Two outcomes per trial: Every single try results in either Success or Failure (like Heads or Tails).
3. Constant probability ($p$): The chance of success must remain identical for every trial (e.g., a 50% chance of heads every single flip).
4. Independence: The result of one trial doesn't influence or change the result of the next trial.

```python
from scipy.stats import binom

# Parameters for our free-throw scenario
n = 5   # Total number of independent trials (n)
p = 0.7 # Success probability per individual trial (p)

# 1. Calculate the exact probability of making exactly 4 out of 5 shots
# PMF stands for Probability Mass Function (finds exact counts)
prob_4 = binom.pmf(k=4, n=n, p=p)
print(f"Exact probability of making 4 shots: {prob_4:.4f} (~{prob_4 * 100:.1f}%)")

# 2. Simulate 1,000 games where the player takes 5 free throws each game
# rvs stands for Random Variates (generates random samples)
simulations = binom.rvs(n=n, p=p, size=1000) # random variates

# Let's check the average makes across our simulation vs. the theoretical expected value (n * p)
average_makes = sum(simulations) / len(simulations)
expected_value = n * p

print(f"Simulated average makes per round: {average_makes:.2f}")
print(f"Theoretical expected value (n * p): {expected_value:.2f}")

```

What this code does:
1. binom.pmf(k=4, n=5, p=0.7) computes the exact math formula we talked about earlier, giving us roughly a $36\%$ chance of making exactly 4 shots.
2. binom.rvs(n=5, p=0.7, size=1000) rolls the virtual dice 1,000 times. When you run it, you'll see that the simulated average comes out extremely close to our theoretical expectation of $3.5$ makes per round ($5 \times 0.7$).

```python
# Set distribution parameters
n_binom = 20  # Total number of independent trials (n)
p_binom = 0.4  # Success probability per individual trial (p)

# Define discrete support domain vector: k ∈ {0, 1, 2, ..., n}
k_binom = np.arange(0, n_binom + 1)

# Compute analytical PMF array P(X = k) across all values of k
pmf_binom = stats.binom.pmf(k_binom, n=n_binom, p=p_binom)

# Perform empirical Monte Carlo sampling (N = 10,000 random variable draws)
simulated_binom = stats.binom.rvs(
    n=n_binom, p=p_binom, size=10000, random_state=42
)

# Display theoretical vs empirical comparison metrics
print("=== 2. BINOMIAL DISTRIBUTION (n=20, p=0.4) ===")
print(f"Theoretical Mean E[X] = n*p:             {n_binom * p_binom:.4f}")
print(f"Empirical Sample Mean:                   {np.mean(simulated_binom):.4f}")
print(f"Theoretical Variance Var(X) = n*p*(1-p): {n_binom * p_binom * (1 - p_binom):.4f}")
print(f"Empirical Sample Variance:               {np.var(simulated_binom):.4f}\n")
```

## 4. Poisson Distribution

### Mathematical Definition
Models the number of independent events $k$ occurring in a fixed interval of time or space, given a constant average rate $\lambda$.

$$X \sim \text{Poisson}(\lambda)$$
- *Pronunciation:* "X follows a Poisson distribution with rate parameter lambda."

#### Probability Mass Function (PMF)
$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}, \quad k \in \{0, 1, 2, \dots\}$$
- *Pronunciation:* "Probability that X equals k is lambda to the power k, times e to the minus lambda, divided by k factorial, for k greater than or equal to 0."
- **Euler's Constant:** $e \approx 2.71828$.
- **Note on Python Execution:** NumPy universal functions (ufuncs) like `np.exp()` accept argument values *positionally* (`np.exp(-lambda_val)`). Passing keyword arguments such as `x=` triggers a `TypeError`.

#### Key Moments
- **Mean:** $\mathbb{E}[X] = \lambda$
- **Variance:** $\text{Var}(X) = \lambda$
  - *Unique Property:* The mean and variance of a Poisson distribution are identical: $\mathbb{E}[X] = \text{Var}(X) = \lambda$.

```python
from scipy.stats import poisson

lambda_rate = 3
prob_2 = poisson.pmf(k=2, mu=lambda_rate)
print(f"Probability of getting exactly 2 orders: {prob_2:.4f} (~{prob_2 * 100:.1f}%)")
simulated_hours = poisson.rvs(mu=lambda_rate, size=10)
print("Simulated orders for 10 hours (random values of X):", simulated_hours)
```

```python
from scipy import special, stats
# Set rate parameter lambda (λ): mean event count per unit interval
lambda_poisson = 5.0  # Average arrival rate (e.g., 5 incoming support tickets per hour)

# Truncated discrete evaluation domain: k ∈ {0, 1, 2, ..., 15}
k_poisson = np.arange(0, 16)

# 1. Compute analytical PMF values using SciPy stats
pmf_poisson = stats.poisson.pmf(k_poisson, mu=lambda_poisson)

# 2. Direct manual formula evaluation demonstrating NumPy positional ufunc rules:
# Formula: P(X = k) = (lambda^k * exp(-lambda)) / k!
# Correct: Pass positional argument '-lambda_poisson' to np.exp()
manual_pmf_poisson = (
    (lambda_poisson**k_poisson) * np.exp(-lambda_poisson)     / special.factorial(k_poisson)  
)
# np.exp(-lambda_poisson) :  # POSITIONAL ARGUMENT passed to np.exp
# stats.factorial(k_poisson)  # Factorial evaluation via scipy.special

# Confirm numerical equivalence between SciPy and direct manual formula computation
assert np.allclose(pmf_poisson, manual_pmf_poisson), "PMF computation mismatch!"

print("=== 3. POISSON DISTRIBUTION (lambda=5.0) ===")
print(f"P(X = 5 arrivals):                   {pmf_poisson[5]:.4f}")
print(f"Cumulative P(X <= 3 arrivals):       {stats.poisson.cdf(k=3, mu=lambda_poisson):.4f}")
print(f"Theoretical Mean = Variance = Lambda: {lambda_poisson:.2f}\n")
```

## 5. Geometric Distribution
### Mathematical Definition
Models the total number of trial attempts $k$ required to observe the **first success** in a sequence of independent Bernoulli trials with constant success probability $p$.

$$X \sim \text{Geometric}(p)$$
- *Pronunciation:* "X follows a Geometric distribution with success parameter p."

#### Probability Mass Function (PMF)
$$P(X = k) = (1 - p)^{k - 1} p, \quad k \in \{1, 2, 3, \dots\}$$
- *Pronunciation:* "Probability that X equals k is 1 minus p to the power k minus 1, times p, for k greater than or equal to 1."
- **Memoryless Property:** $P(X > s + t \mid X > s) = P(X > t)$.
  - *Pronunciation:* "Probability that X is greater than s plus t given X is greater than s equals probability that X is greater than t."
  - *Meaning:* The probability of achieving success on future trials does not depend on past failures.

#### Key Moments
- **Mean:** $\mathbb{E}[X] = \frac{1}{p}$
  - *Pronunciation:* "Expected value of X equals 1 over p."
- **Variance:** $\text{Var}(X) = \frac{1 - p}{p^2}$
  - *Pronunciation:* "Variance of X equals 1 minus p divided by p squared."
