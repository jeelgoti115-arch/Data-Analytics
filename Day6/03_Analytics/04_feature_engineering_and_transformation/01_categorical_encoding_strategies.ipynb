```python
# Global Imports and Runtime Visual Configuration
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visual styling settings
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 120
plt.rcParams["figure.figsize"] = (8, 4)
plt.rcParams["axes.titlesize"] = 12
plt.rcParams["axes.labelsize"] = 10

print("Global libraries for categorical encoding strategies loaded successfully.")
```

```python
df_cat = pd.DataFrame({'color': ['Red', 'Blue', 'Green', 'Blue', 'Red']})
df_one = pd.get_dummies(df_cat,prefix='color',dtype=int,drop_first=True)
print(df_one)
```

```python
df_ord = pd.DataFrame({'size': ['Small', 'Medium', 'Large', 'Medium', 'Small']})
mapping = {'Small': 1, 'Medium': 2, 'Large': 3}
df_ord['size_encoded'] = df_ord['size'].map(mapping)
print("Ordinal Encoded DataFrame:\n", df_ord)
```

```python
## 3. Target / Mean Encoding with Smoothing
df_target = pd.DataFrame({
    'city': ['NY', 'SF', 'NY', 'CHI', 'SF'],
    'target': [1, 0, 1, 0, 1]
})
mean_enc = df_target.groupby('city')['target'].mean().to_dict()
print(mean_enc)
df_target['city_encoded'] = df_target['city'].map(mean_enc)
print("Target Encoded Mapping Result:\n", df_target)
```

```python
## 4. Handling High-Cardinality Categoricals

df_high = pd.DataFrame({'sku': ['A','A', 'B', 'B','C', 'Rare1', 'Rare2', 'Rare3']})
counts = df_high['sku'].value_counts()
frequent_skus = counts[counts > 1].index
df_high['sku_cleaned'] = df_high['sku'].apply(lambda x: x if x in frequent_skus else 'Other')
print(
    "High-Cardinality Collapsed Result:\n",
    df_high,
)
```

## Summary Table

| Encoding Strategy | Core Mechanism | Dimensionality Impact | Best Suited For |
| :--- | :--- | :--- | :--- |
| **One-Hot / Dummy Encoding** | Binary indicator columns per level | Expands width ($K$ columns) | Nominal categories with low-to-moderate cardinality |
| **Ordinal & Frequency Encoding** | Integer rank or occurrence ratio | Compact (1 column) | Ordered categories or frequency representation |
| **Target / Mean Encoding** | Expected target value with smoothing | Compact (1 column) | High-cardinality features in boosted trees or linear models |
| **High-Cardinality Handling** | Frequency thresholding & feature hashing | Controlled dimension size | Millions of unique tokens, IP addresses, or SKUs |
