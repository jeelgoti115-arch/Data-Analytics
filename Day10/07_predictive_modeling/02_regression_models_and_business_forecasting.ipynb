```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visual styling settings
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 120
plt.rcParams["figure.figsize"] = (8, 4)
plt.rcParams["axes.titlesize"] = 12
plt.rcParams["axes.labelsize"] = 10

print("Global libraries for regression models and business forecasting loaded successfully.")
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_n x_n$$

$$y = \beta_0 + \beta_1 x$$
- $y$ = The target you want to predict (e.g., house price).
- $x$ = The input feature (e.g., square footage).
- $\beta_0$ (Intercept / Bias): The expected value of $y$ when all input features ($x$) are zero. (Where the line crosses the vertical Y-axis).
- $\beta_1$ (Coefficient / Slope): How much $y$ changes for every 1-unit increase in $x$.

```python
# y = mx + c
# y = target
# x = input feature
# c = intercept/bias
# m = slope/coefficient
'''

'''
np.random.seed(42)
X_reg = pd.DataFrame({
    'ad_spend': np.random.uniform(10, 100, 100),
    'store_size': np.random.uniform(500, 3000, 100)
})
y_reg = 50 + 2.5 * X_reg['ad_spend'] + 0.05 * X_reg['store_size'] + np.random.normal(0, 5, 100)

lin_reg = LinearRegression()
lin_reg.fit(X_reg, y_reg)

df_coef = pd.DataFrame({'Feature': X_reg.columns, 'Coefficient': lin_reg.coef_})
print(
    "Linear Regression Intercept:",
    round(lin_reg.intercept_, 2),
)
print("Feature Coefficients:\n", df_coef.round(4))
```

#### 2. Regularization (Ridge and Lasso) for Feature Selection

##### 1. Definition
Regularization adds penalty terms to regression loss functions to prevent overfitting;
- Ridge ($\ell_2$) shrinks coefficients toward zero, while 
- Lasso ($\ell_1$) can drive irrelevant feature coefficients exactly to zero for automated feature selection.

### 3. Notation Pronunciation
- `Ridge()`: "L2-norm penalized linear regression"
- `Lasso()`: "L1-norm penalized sparse feature selection regressor"
- $\alpha$: "Regularization strength penalty multiplier"

### 4. Naive Example
Using Lasso regression to eliminate 15 redundant customer demographic columns while retaining only the 3 most predictive variables.
### 5. Uses
- Mitigating multicollinearity in high-dimensional datasets.
- Automated feature selection and model parsimony.
- Improving out-of-sample generalization.

```python
ridge = Ridge(alpha=1.0)
lasso = Lasso(alpha=0.1)
ridge.fit(X_reg, y_reg)
lasso.fit(X_reg, y_reg)

df_reg_comp = pd.DataFrame({
    'Feature': X_reg.columns,
    'Ridge_Coef': ridge.coef_,
    'Lasso_Coef': lasso.coef_
})
print("Regularized Coefficients Comparison:\n", df_reg_comp.round(4))

'''
When comparing Ridge and Lasso coefficients, we are checking model stability. 
If both methods give us nearly identical, non-zero numbers for our features, 
it proves our features are robust, independent, and genuinely driving 
the prediction without overfitting
'''
```

#### 3. Evaluation Metrics (RMSE, MAE, R-Squared, MAPE)

##### 1. Definition
Regression evaluation metrics quantify forecasting accuracy; 
- MAE measures average absolute error, 
- RMSE penalizes large errors heavily, 
- $R^2$ measures variance explained, and 
- MAPE computes percentage error.

##### 3. Notation Pronunciation
- `RMSE`: "Root mean squared error"
- `MAE`: "Mean absolute error"
- `R-Squared`: "Coefficient of determination ($0$ to $1$)"

##### 4. Naive Example
Reporting an average sales prediction error of \$1,200 (MAE) with an $R^2$ of 0.82, indicating the model explains 82% of revenue variance.
##### 5. Uses
- Assessing and comparing regression model performance.
- Communicating forecast accuracy to executive stakeholders.
- Guiding hyperparameter tuning.

## Regression Evaluation Metrics Reference

### Quick Comparison Table

| Metric | Full Name | Unit | Sensitivity to Outliers | Best Used For |
| :--- | :--- | :--- | :--- | :--- |
| **MAE** | Mean Absolute Error | Same as target | Low (Treats all errors equally) | General business interpretation where outliers shouldn't dominate. |
| **MSE** | Mean Squared Error | Squared ($y^2$) | High (Punishes large errors heavily) | Mathematical optimization; penalizing big mistakes. |
| **RMSE** | Root Mean Squared Error | Same as target | High (Punishes large errors heavily) | Standard default metric; penalizing large misses while keeping original units. |
| **MAPE** | Mean Absolute Percentage Error | Percentage (%) | Moderate | Comparing performance across datasets with different scales (e.g., store revenue). |
| **$R^2$** | Coefficient of Determination | Unitless (0 to 1) | Moderate | Explaining goodness-of-fit (how much variance your model captures). |

---

### 1. MAE (Mean Absolute Error)
* **What it measures:** The average absolute distance between your predictions and the actual values.
* **Formula concept:** $\text{Average of } |y_{\text{actual}} - y_{\text{predicted}}|$
* **Interpretation:** If you are predicting house prices and your MAE is \$15,000, it means your model's predictions are, on average, off by \$15,000.
* **Key Takeaway:** It treats all errors linearly. A 10-unit error is weighted exactly twice as much as a 5-unit error. It is robust to outliers.

### 2. MSE (Mean Squared Error)
* **What it measures:** The average of the *squared* differences between predictions and actual values.
* **Formula concept:** $\text{Average of } (y_{\text{actual}} - y_{\text{predicted}})^2$
* **Interpretation:** Because the errors are squared, a prediction that is off by 10 units results in an error score of $100$, while a 2-unit error is only $4$.
* **Key Takeaway:** It heavily penalizes large errors. However, because the units are squared (e.g., $\text{dollars}^2$), it is hard to interpret intuitively on its own.

### 3. RMSE (Root Mean Squared Error)
* **What it measures:** The square root of the MSE.
* **Formula concept:** $\sqrt{\text{MSE}}$
* **Interpretation:** It brings the squared error back down into the original units of your target variable. If your target is temperature in Celsius, RMSE is also in Celsius.
* **Key Takeaway:** RMSE is one of the most popular regression metrics. Like MSE, it severely penalizes large errors, meaning a few massive mistakes will drive your RMSE up significantly, even if your other predictions are spot on.

### 4. MAPE (Mean Absolute Percentage Error)
* **What it measures:** The average percentage error of your predictions.
* **Formula concept:** $\text{Average of } \left| \frac{y_{\text{actual}} - y_{\text{predicted}}}{y_{\text{actual}}} \right| \times 100\%$
* **Interpretation:** A MAPE of 5% means your model is off by 5% on average relative to the actual values.
* **Key Takeaway:** Highly intuitive for business stakeholders ("we are off by 8% on sales forecasts"). **Caution:** It breaks down or gives infinite values if any of your actual target values ($y_{\text{actual}}$) are zero or close to zero.

### 5. $R^2$ (R-Squared / Coefficient of Determination)
* **What it measures:** How well your model explains the variance of the target data compared to a simple baseline (just guessing the average value for everything).
* **Interpretation:**
  * **$1.0$:** Perfect predictions (100% of variance explained).
  * **$0.0$:** Your model is no better than just predicting the average of the target column.
  * **Negative:** Your model is worse than a horizontal baseline line.
* **Key Takeaway:** Unlike MAE or RMSE (which depend on your data's scale), $R^2$ is a standardized score from roughly $0$ to $1$, making it easy to judge overall model fit across different projects.

---

### Summary Rule of Thumb
* Want to know the typical error in plain units? Use **MAE** or **RMSE**.
* Want to heavily punish catastrophic mistakes? Use **RMSE** (or MSE).
* Want to communicate error as a percentage to non-technical stakeholders? Use **MAPE**.
* Want to know what percentage of data patterns your model actually captured? Use **$R^2$**.

```python
preds_reg = lin_reg.predict(X_reg)
mae = mean_absolute_error(y_reg, preds_reg)
rmse = np.sqrt(mean_squared_error(y_reg, preds_reg))
r2 = r2_score(y_reg, preds_reg)

df_metrics = pd.DataFrame({
    'Metric': ['MAE', 'RMSE', 'R-Squared'],
    'Value': [mae, rmse, r2]
})
print("Regression Evaluation Metrics:\n", df_metrics.round(3))
```

## 4. Residual Diagnostics for Business Predictions

### 1. Definition
Residual diagnostics evaluate model error patterns (residuals) against predicted values to check linear regression assumptions such as homoscedasticity, normality, and absence of non-linear trends.

### 2. Math Formula
Residual error $e_i$ for sample $i$ computed as actual minus prediction:
$$e_i = y_i - \hat{y}_i$$

### 3. Notation Pronunciation
- `Residual Error`: "Difference between actual value and model prediction"
- `Homoscedasticity`: "Constant variance of residual errors across predicted values"

### 4. Naive Example
Plotting residuals and noticing prediction errors increase for larger stores, indicating heteroscedasticity that requires feature transformation.
### 5. Uses
- Validating linear regression assumptions before business deployment.
- Detecting outliers and systematic model bias.
- Improving model robustness.

```python
residuals = y_reg - preds_reg
df_resid = pd.DataFrame({"atcual":y_reg,'predicted': preds_reg, 'residual': residuals})
print(df_resid.head())
print("=" * 60)
print(
    "Residual Summary Statistics:\n",
    df_resid['residual'].describe().round(3),
)
```
