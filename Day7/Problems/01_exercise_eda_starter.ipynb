````python
# Setup & Data Generation
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)

# Seed for reproducibility
np.random.seed(2026)
n_samples = 300

# Synthetic Telecom Customer Churn Dataset
tenure_months = np.random.randint(1, 72, size=n_samples)
monthly_charges = np.random.normal(loc=65, scale=20, size=n_samples).round(2)
total_charges = (tenure_months * monthly_charges) + np.random.normal(0, 100, size=n_samples)
total_charges = np.maximum(20, total_charges).round(2)

contract_type = np.random.choice(["Month-to-Month", "One-Year", "Two-Year"], size=n_samples, p=[0.55, 0.25, 0.20])
tech_support = np.random.choice(["Yes", "No"], size=n_samples, p=[0.4, 0.6])

# Target variable: Churn likelihood driven by contract type and charges
churn_prob = np.where(contract_type == "Month-to-Month", 0.4, 0.1)
churn = np.random.binomial(1, churn_prob)

df_churn = pd.DataFrame({
    "tenure_months": tenure_months,
    "monthly_charges": monthly_charges,
    "total_charges": total_charges,
    "contract_type": contract_type,
    "tech_support": tech_support,
    "churn": churn
})

# Inject missing values for missingness analysis practice
df_churn.loc[df_churn.sample(frac=0.08, random_state=42).index, "total_charges"] = np.nan
df_churn.loc[df_churn.sample(frac=0.05, random_state=42).index, "tech_support"] = np.nan

print("--- Telecom Churn Dataset Preview ---")
print(df_churn.head())
````
